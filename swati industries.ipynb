```python
sns.countplot(x = 'material_name', data = df)
```

```python
ax = sns.countplot( x = 'material_name', data = df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df.groupby([ 'material_name'], as_index= False)['cost_per_kg_inr'].sum().sort_values(by='cost_per_kg_inr', ascending= False)
```

```python
sales_gen = df.groupby([ 'material_name'], as_index= False)['quantity_used_kg'].sum().sort_values(by='quantity_used_kg', ascending= False)
sns.barplot(x = 'material_name', y = 'quantity_used_kg', data = sales_gen)
```
